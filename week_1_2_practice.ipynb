```python
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
```

```python
df = yf.download('EURUSD=X', period='1mo', interval='1h')
df = df[['Close']].reset_index()
print(df.head())
```

```python
df.dropna(inplace=True)
print("Rows after cleaning:", len(df))
```

```python
plt.figure(figsize=(8, 4))
plt.plot(df['Datetime'], df['Close'], color='blue')
plt.title('EURUSD Close (1 month)')
plt.xlabel('Datetime')
plt.ylabel('Price')
plt.tight_layout()
plt.show()
```

```python
df.to_csv('eurusd_cleaned.csv', index=False)
print("CSV saved as eurusd_cleaned.csv")
```
